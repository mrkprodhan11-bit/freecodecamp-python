# **A university offers a special programming workshop to students who satisfy certain eligibility conditions.**
```
A student is eligible if:

their programming score is 60 or higher AND their attendance is 75% or higher, OR
they have a valid scholarship.

Write a Python program that takes the student's programming score, attendance percentage, and scholarship status as input and determines whether the student is eligible.

The scholarship status will be entered as:

1 → Has scholarship
0 → Does not have scholarship

Your program should print:

Eligible for the workshop

if the student is eligible.

Otherwise, print:

Not eligible for the workshop
Sample Input 1
Please enter programming score: 70
Please enter attendance: 80
Please enter scholarship status: 0
Sample Output 1
Eligible for the workshop
Sample Input 2
Please enter programming score: 45
Please enter attendance: 90
Please enter scholarship status: 0
Sample Output 2
Not eligible for the workshop

````python
score = float(input("please enter programming score: "))
attendence = float(input("please enter the percentage of your attendence: "))
scholarship_status = int(input("please enter the scholarship status: "))
if score >= 60 and attendence >= 75 or scholarship_status == 1:
  print("Eligible for the workshop")
else:
  print("Not eligible for the workshop")
````

# **📦 Delivery Charge Calculator**

```
A worldwide delivery company calculates the delivery charge based on the package weight and delivery zone.

The program takes:

weight — package weight in kg
zone
1 → Local
2 → International
express
1 → Express delivery
0 → Normal delivery
Rules

For Local delivery:

Weight up to and including 2 kg → 300 BDT
Weight above 2 kg → 300 + 100 BDT for every additional kg

For International delivery:

Weight up to and including 2 kg → 1000 BDT
Weight above 2 kg → 1000 + 500 BDT for every additional kg

If express delivery is selected, add an additional 300 BDT to the calculated charge.

Important

If the entered weight is 0 or negative, print:

Invalid Weight

and do not calculate the delivery charge.

Output

For a valid package, print:

Delivery Charge: X BDT
Sample Input 1
Please enter package weight: 2
Please enter delivery zone: 1
Please enter express status: 0
Sample Output 1
Delivery Charge: 300 BDT
Sample Input 2
Please enter package weight: 4
Please enter delivery zone: 1
Please enter express status: 1
Sample Output 2
Delivery Charge: 800 BDT

Calculation:

300 + (2 × 100) + 300 = 800

Sample Input 3
Please enter package weight: 3
Please enter delivery zone: 2
Please enter express status: 0
Sample Output 3
Delivery Charge: 1500 BDT
Sample Input 4
Please enter package weight: -2
Please enter delivery zone: 1
Please enter express status: 1
Sample Output 4
Invalid Weight

````python
weight = float(input("Please enter package weight: "))
zone = int(input("Please enter delivery zone: "))
express = int(input("Please enter express status: "))

if weight <= 0:
    print("Invalid Weight")
else:
    if zone == 1:
        if weight <= 2:
            charge = 300
        else:
            charge = 300 + (weight - 2) * 100
    elif zone == 2:
        if weight <= 2:
            charge = 1000
        else:
            charge = 1000 + (weight - 2) * 500

    if express == 1:
        charge = charge + 300

    print(f"Delivery Charge: {charge:g} BDT")
````
